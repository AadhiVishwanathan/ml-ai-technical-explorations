```python
driver_code = '''
#include "optimization/multi_objective/subpopulation_framework/Subpopulation_Framework.h"
#include "optimization/multi_objective/problems/WFG_Problem.h"
#include "random/State_of_Art_Random.h"
#include "time.h"
#include "stdio.h"

int main()
{
    double max_limit = 1;
    double min_limit = 0;
    int population_size = 3;
    int subpopulation_size = 20;

    int position_parameter = 4;
    int distance_parameter = 20;
    int number_of_problems = 2;  // 2-objective WFG (matches paper's 2-objective tests)
    int generations = 2000;      // reduced from paper's likely larger value, for feasible runtime

    float F = 0.1;
    float CR = 0.1;

    printf("Running SAN (Subpopulation Framework) on WFG1, %d objectives, %d generations\\n",
           number_of_problems, generations);

    Random* random = new State_of_Art_Random(time(NULL));
    Optimization_Problem* problem = new WFG_Problem(WFG1, position_parameter, distance_parameter, number_of_problems);

    Subpopulation_Framework* san = new Subpopulation_Framework(
        generations, population_size, subpopulation_size, max_limit, min_limit, F, CR, random
    );

    double* solution;
    double result = san->optimize(problem, number_of_problems, &solution);
    printf("Result: %f\\n", result);

    san->storeSolutions("san_wfg1_solutions.txt");
    printf("Solutions saved.\\n");

    return 0;
}
'''

with open('optimization/multi_objective/subpopulation_framework/san_driver.cpp', 'w') as f:
    f.write(driver_code)

print("Driver file written")
```

```python
def pareto_front(points):
    """Points: assume minimization. Return boolean mask of non-dominated points."""
    is_efficient = np.ones(points.shape[0], dtype=bool)
    for i, c in enumerate(points):
        if is_efficient[i]:
            is_efficient[is_efficient] = np.any(points[is_efficient] < c, axis=1) | np.all(points[is_efficient] == c, axis=1)
            is_efficient[i] = True
    return is_efficient

objectives = valid_data[:, -2:]
pareto_mask = pareto_front(objectives)
pareto_points = objectives[pareto_mask]
print(f"Pareto-optimal solutions: {pareto_mask.sum()} out of {len(objectives)}")

plt.figure(figsize=(8,6))
plt.scatter(objectives[:,0], objectives[:,1], s=8, alpha=0.3, label='All solutions')
plt.scatter(pareto_points[:,0], pareto_points[:,1], s=25, color='red', label='Pareto front')
plt.xlabel('Objective 1')
plt.ylabel('Objective 2')
plt.title('SAN — Pareto Front on WFG1')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
def hypervolume_2d(points, reference_point):
    """Simple 2D hypervolume calculation for minimization (points should be sorted)."""
    sorted_points = points[points[:, 0].argsort()]
    hv = 0
    prev_x = reference_point[0]
    for point in sorted_points:
        if point[0] < prev_x:
            width = prev_x - point[0]
            height = reference_point[1] - point[1]
            if height > 0:
                hv += width * height
            prev_x = point[0]
    return hv

# Reference point: worse than any observed objective value (standard practice: slightly beyond the max)
ref_point_wfg1 = [pareto_points[:,0].max() + 0.1, pareto_points[:,1].max() + 0.1]
ref_point_wfg2 = [pareto_points_wfg2[:,0].max() + 0.1, pareto_points_wfg2[:,1].max() + 0.1]

hv_wfg1 = hypervolume_2d(pareto_points, ref_point_wfg1)
hv_wfg2 = hypervolume_2d(pareto_points_wfg2, ref_point_wfg2)

print(f"WFG1 Hypervolume: {hv_wfg1:.4f} (reference point: {ref_point_wfg1})")
print(f"WFG2 Hypervolume: {hv_wfg2:.4f} (reference point: {ref_point_wfg2})")
```
